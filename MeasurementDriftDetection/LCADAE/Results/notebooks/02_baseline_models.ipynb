# 02 — Manuscript Baseline Experiments

This notebook trains published baselines with independent random seeds and evaluates them on:

- the balanced severity–duration scenario grid;
- the sparse mixed-fault test;
- natural Gas Sensor batch shift;
- native SKAB anomaly sequences;
- a validation-only threshold sensitivity grid.

Duplicate experiment keys are skipped automatically.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT / "src"))

import numpy as np
import pandas as pd
from IPython.display import Image, Markdown, display

from drift_detection.config import load_config

config = load_config()
print(f"Project root: {PROJECT_ROOT}")
print(f"Experiment version: {config['project']['experiment_version']}")
print(f"Scenario profile: {config['experiment']['profile']}")

import importlib.util
from drift_detection.experiments_v2 import manuscript_result_root, run_baseline_manuscript_suite
```

## Controls

For final manuscript results, use all configured seeds. Install `requirements-sota.txt` before enabling published SOTA comparators.


```python
SEEDS = list(config["project"]["seed_list"])
FORCE_RERUN = False
DEEPOD_AVAILABLE = importlib.util.find_spec("deepod") is not None
RUN_OPTIONAL_SOTA = DEEPOD_AVAILABLE

print("Independent seeds:", SEEDS)
print("DeepOD available:", DEEPOD_AVAILABLE)
print("Optional SOTA enabled:", RUN_OPTIONAL_SOTA)
if not DEEPOD_AVAILABLE:
    print("Install requirements-sota.txt and rerun this notebook before manuscript submission.")
```

## Required benchmark files


```python
required = [
    config["paths"]["benchmarks"] / "gas_controlled.npz",
    config["paths"]["benchmarks"] / "skab_controlled.npz",
    config["paths"]["benchmarks"] / "gas_sensor_scenario_manifest.csv",
    config["paths"]["benchmarks"] / "skab_scenario_manifest.csv",
]
missing = [str(path) for path in required if not path.exists()]
if missing:
    raise FileNotFoundError("Run Notebook 01 first:\n" + "\n".join(missing))
```

## Gas Sensor baselines


```python
gas_baselines = run_baseline_manuscript_suite(
    config["paths"]["benchmarks"] / "gas_controlled.npz",
    dataset="gas_sensor",
    config=config,
    seeds=SEEDS,
    run_optional_sota=RUN_OPTIONAL_SOTA,
    force=FORCE_RERUN,
)
display(gas_baselines[[column for column in ["model", "seed", "status", "f1", "auprc", "event_recall", "false_positive_rate"] if column in gas_baselines.columns]])
```

## SKAB baselines


```python
skab_baselines = run_baseline_manuscript_suite(
    config["paths"]["benchmarks"] / "skab_controlled.npz",
    dataset="skab",
    config=config,
    seeds=SEEDS,
    run_optional_sota=RUN_OPTIONAL_SOTA,
    force=FORCE_RERUN,
)
display(skab_baselines[[column for column in ["model", "seed", "status", "f1", "auprc", "event_recall", "false_positive_rate", "native_f1"] if column in skab_baselines.columns]])
```

## Audit the new baseline records


```python
result_dir = manuscript_result_root(config) / "baselines"
for file_name in [
    "runs.csv",
    "fault_summary.csv",
    "threshold_sensitivity.csv",
    "gas_natural_metrics.csv",
    "skab_native_metrics.csv",
]:
    path = result_dir / file_name
    if path.exists():
        print(f"\n{file_name}: {path}")
        display(pd.read_csv(path).tail(20))
```

## Distinct-seed check


```python
runs = pd.read_csv(result_dir / "runs.csv")
completed = runs[runs["status"] == "completed"]
seed_audit = completed.groupby(["dataset", "model"], as_index=False)["seed"].nunique().rename(columns={"seed": "distinct_seed_count"})
display(seed_audit)
expected = len(SEEDS)
if (seed_audit["distinct_seed_count"] < expected).any():
    print("Some models are incomplete. Inspect failed rows before aggregation.")
else:
    print("All completed baseline groups contain the requested independent seeds.")
```
